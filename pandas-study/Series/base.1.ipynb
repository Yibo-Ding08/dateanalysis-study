### 创建

```python
# series的创建
import pandas as pd
s=pd.Series([1, 2, 3, 4, 5])
print(s)#左索引,右数值
#自定义索引
s=pd.Series([10, 2, 30, 40, 5],index=['a','b','c','d','e'])#改索引
print(s)
s=pd.Series([10, 2, 30, 40, 5],index=['a','b','c','d','e'],name="月份")#表示这一列是做什么的
print(s)
```

```python
#通过字典来创建
s=pd.Series({'a':10,'b':20,'c':30})
print(s)
s1=pd.Series(s,index=['a','d'])#d没有值,所以是nan
print(s1)
```
