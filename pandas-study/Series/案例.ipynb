```python
"""
创建一个包含10名学生数学成绩的series,
成绩范围在50-100人之间,
计算平均分,最高分,最低分,并找出高于平均分的人数
"""
import pandas as pd
import numpy as np
np.random.seed(0)
scores=pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)])
print(scores)
print("平均分:", scores.mean())
print(f"最高分: {scores.max()}, 最低分: {scores.min()}")
print(f"高于平均分的人数:{scores[scores>scores.mean()].count()}")
print(f"高于平均分的学生名单:{scores[scores>scores.mean()].index.tolist()}")
```

```python
"温度数据分析"
from operator import index

import pandas as pd
import numpy as np
tem=pd.Series([28,31,29,32,30,27,33], index=['周一','周二','周三','周四','周五','周六','周日'])
print(f"大于30度的天数:{tem[tem>30].count()}")
print(tem.sort_values(ascending=False))
print(f"平均值为{tem.mean()}")
t1=tem.diff().abs()
t1.sort_values(ascending=False).index[0:2]
print(f"温度变化最大的两天为:{t1.sort_values(ascending=False).index[0:2].tolist()}")
```

```python
"股票价格分析"
prices=pd.Series([102.3,103.5,104.2,105.1,106.8,107.9,108.6,109.4,110.2,109.5],index=pd.date_range('2023-01-01',periods=10))
print(prices)
print(f"每天收益率为:{prices.pct_change()}")#percent change
t=prices.pct_change()
print(f"找出收益最高和最低的日期,最高:{t.idxmax()},最低:{t.idxmin()}")
print(f"标准差:{t.std()}")
```
