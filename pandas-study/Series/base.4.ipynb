### 常用方法

```python
import pandas as pd
s=pd.Series([10,2,None,5,5],index=['a','b','c','d',"e"],name="test")
print(s.head())
s.tail(2)
s.describe()#描述性统计
```

```python
#获得元素个数
s.count()
```

```python
#获取索引
print(s.index)
print(s.keys())
```

```python
#是否为缺失值
print(s.isna())
s.isna()
s.isin([10,2,5])#判断元素是否在列表中
```

```python
print(s.mean())#平均值
print(s.median())#中位数
print(s.sum())#总和
print(s.max())#最大值
print(s.var())#方差
```

```python
s.sort_values()#排序
print(s.sort_values())
print(s.quantile(0.9))#分位数
print(s.value_counts())#出现次数
```

```python
s.drop_duplicates()#去重
s.unique()#去重,返回数 组
```

```python
print(s.nunique())#去重,返回个数
```

```python
s.sort_index()
s.sort_values()
```
