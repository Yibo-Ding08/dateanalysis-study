### 常用方法

```python
import pandas as pd
s=pd.Series([10,2,None,5],index=['a','b','c','d'],name="test")
print(s.head())
s.tail(2)
s.describe()#描述性统计
```

```python
#获得元素个数
s.count()
```

```python
#获取索引
print(s.index)
print(s.keys())
```

```python
#是否为缺失值
print(s.isna())
s.isna()
s.isin([10,2,5])#判断元素是否在列表中
```
