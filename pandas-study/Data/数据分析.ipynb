```python
#分组聚合
#df.groupby("分组字段")['聚合字段]
from numpy import astype
import pandas as pd
df=pd.read_csv('employees.csv')
# df['department_id']=df['department_id'].astype('int64')
# df
#df['department_id']=df['department_id'].replace('NAN',pd.NA)
print(df['department_id'].isna().sum())
df.dropna(subset=['department_id'],inplace=True)
df['department_id']=df['department_id'].astype('int64')
df.groupby('department_id').groups#分组
df.groupby('department_id').get_group(30)#分组查找
df.groupby('department_id')['salary'].mean()#分组求平均工资
df2=df.groupby('department_id')[['salary']].mean()
df2['salary']=df2['salary'].round(3)
df2.reset_index()
df2.sort_values('salary',ascending=False)
df


```

```python
#计算不同部门不同岗位的人的平均薪资
df2=df.groupby(['department_id','job_id'])[['salary']].mean()
df2.reset_index()
df2.sort_values('salary',ascending=False)
```

```python
#企鹅数据分析
#1.导入库,导入数据
#2.数据清洗
#3.数据特征的构造
#4.数据分析

import pandas as pd
df=pd.read_csv('penguins.csv')
df.isna().sum()
df.dropna(inplace=True)
df.isna().sum()
df['sex']=df['sex'].astype('category')
df.info()
df['bill_ratio']=df['bill_length_mm']/df['bill_depth_mm']
df['mass_level']=pd.cut(df['body_mass_g'],bins=3,labels=['低','中','高'])
print(df['mass_level'].value_counts())
#按岛屿和性别分组
df.groupby(['sex']).agg({
    'body_mass_g':['mean','count']
})
df.groupby(['island','sex']).agg({
    'body_mass_g':['mean','count']
})


```

```python
from holoviews import Labels
from httpx import head
from numpy import mean
import pandas as pd
df=pd.read_csv('sleep.csv')
df.describe()
df.isna().sum()
df.drop(columns='sleep_disorder',inplace=True)
df['gender']=df['gender'].astype('category')
df.info()
df['occupation']=df['occupation'].astype('category')
df.head()
df[['high','low']]=df['blood_pressure'].str.split('/',expand=True).astype('int64')
Labels=['差','中','优']
df['quality_level']=pd.cut(df['sleep_quality'],bins=3,labels=Labels)
df.head()
age_labels=['青少年','中年','老年']
df['age.level']=pd.cut(df['age'],bins=3,labels=age_labels)
df.head()
print(df['bmi_category'].value_counts())
df.groupby(['age.level','bmi_category']).agg({
    'sleep_duration':'mean',
    'sleep_quality':'mean'
})
```
