### 创建

```python
#创建
import pandas as pd
import numpy as np
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([1,2,4,6,7])
df=pd.DataFrame({"第一列":s1,"第二列":s2})
df
df=pd.DataFrame({
   "id":[1,2,3,4,5],
   "name":["张三","李四","王五","赵六","钱七"],
   "age":[18,21,20,21,22], 
},index=["a","b","c","d","e"],columns=["id","age","name"])
df
```

### 属性
**index**:行索引,**values**:值
**dtypes**:元素类型,**shape**:形状
**ndim**:维度,**size**:元素个数
**columns**:列标签,**loc[]**:显示索引
**iloc[]**:隐式索引,**at[]**:按行列标签访问
**iat**:按行列位置访问,**T**:行列转置

```python
#属性
print(df.shape) #行列数
print(df.ndim) #维度
print(df.size) #元素个数
print(df.index) #索引
print(df.columns) #列名
print(df.values) #值,按行来
print(df.dtypes) #数据类型
```

```python
print(df.loc["a":"c"]) #前3行
print(df.iloc[0:3]) #前3行
```

```python
#单个元素
print(df.at["a", "name"]) #张三
print(df.iat[0, 2]) #张三
print(df.loc["a", "name"]) #张三
print(df.iloc[0, 1]) #张三
```

```python
#单列数据
print(df["name"]) #单列数据
```

### 访问数据

```python
#布尔索引筛选数据
df.age>18
df[(df.age>18)&(df.age<21)]
```

```python
#随机取样
df.sample(n=3) #随机取样3行
```

```python
print(df.head(3)) #前3行
print(df.tail(3)) #后3行
print(df.isin(["张三",18])) 
print(df.isna())#是否为空
print(df.sum())#全加
print(df["name"].sum())#name列加
print(df.age.max())
print(df.age.mode())
```

```python
print(df.describe())
```

```python
print(df.count())#每一列非缺失值个数
print(df.value_counts())#出现次数
print(df.duplicated(subset=["age"]))#检查重复
```
