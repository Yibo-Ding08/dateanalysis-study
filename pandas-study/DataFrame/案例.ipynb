```python
"学生成绩分析"
import pandas as pd
data={
    "姓名":['张三','李四','王五','赵六',"钱七"],
    "数学":[85,92,78,88,95],
    "英语":[90,88,85,92,80],
    "物理":[75,80,88,85,90]
}
scores=pd.DataFrame(data)
scores['总分']=scores[['数学','英语','物理']].sum(axis=1)

scores['平均分']=scores['总分']/3
print(scores[(scores["数学"]>90)|(scores['英语']>85)])
scores.sort_values('总分',ascending=False).head(3)


```

```python
"销售数据分析"
import pandas as pd
date={
    "产品名称":["A",'B','C','D'],
    '单价':[100,150,200,120],
    "销售":[50,30,20,40]
}
s=pd.DataFrame(date)
s["总销售额"]=s['单价']*s["销售"]
#s.sort_values('销售',ascending=False).head(1)
s.sort_values("销售",ascending=False)
s.nlargest(1,columns=['总销售额'])
```
