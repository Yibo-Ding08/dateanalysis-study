# 一.ndarray
## 1.ndarray的特性
### 多维性

```python
import numpy as np
```

```python
arr=np.array([1,2,3])
print(arr.ndim)#ndim指数组维度
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2.ndim)
print(arr2)
```

### 同质性

```python
arr=np.array([1,"qw"])#不同的数据类型会被强制转换为相同数据类型
print(arr)
arr3=np.array([1,1.5])
print(arr3)
```

## 2.ndarray的属性

```python
arr=np.array(1)
print(arr)
print("形状:",arr.shape)
print("维度:",arr.ndim)
print("元素个数",arr.size)
print("元素类型:",arr.dtype)
print("元素转置:",arr.T)
```

```python
arr=np.array([1,2,3.5])
print(arr)
print("形状:",arr.shape)
print("维度:",arr.ndim)
print("元素个数",arr.size)
print("元素类型:",arr.dtype)
print("元素转置:",arr.T)
```

```python
arr=np.array([[1,2,3],[1,2,3]])
print(arr)
print("形状:",arr.shape)
print("维度:",arr.ndim)
print("元素个数",arr.size)
print("元素类型:",arr.dtype)
print("元素转置:",arr.T)
```

## 3.ndarray的创建

```python
#基础创建方法
list1=[4,5,6]
arr=np.array(list1,dtype=np.float64)#方法
print(arr.ndim)#属性
print(arr)
```

```python
#copy
arr1=np.copy(arr)#元素跟原始的数组相同,但不是一个数组
print(arr1)
arr1[0]=8
print(arr1)
print(arr)
```

```python
#预定义形状,全0,全1,未初始化,固定值
#全0
arr=np.zeros((2,3),dtype=int)
print(arr)
print(arr.dtype)
arr=np.zeros((2,),dtype=int)
print(arr)

#全1
arr=np.ones((2,1))
print(arr)
#未初始化
arr=np.empty((1,2))
print(arr)
arr=np.full((3,4),2025)
print(arr)
arr1=np.zeros_like(arr)
print(arr1)
```

```python
#等差数列
arr=np.arange(1,10,1)#start,end,step
print(arr)
```

```python
#等间隔数列
app=np.linspace(1,10,9)#start,end,num
print(app)
```

```python
#对数间隔数列
arr=np.logspace(0,4,3,base=2)#start,end,num,底数,底数的几次方
print(arr)
```

```python
#特殊矩阵
#单位矩阵:
arr=np.eye(3,4,dtype=int)#生成单位矩阵
print(arr)
```

```python
#对角矩阵
arr=np.diag([1,2,3])#生成对角矩阵
print(arr)
```

```python
#随机数组的生成
#生成0-1的随机浮点数(分布均匀)
arr=np.random.rand(2,3)
print(arr)
```

```python
#生成指定范围之间的随机浮点数
arr=np.random.uniform(3,6,(2,3))
print(arr)
```

```python
#生成随机整数
arr=np.random.randint(3,6,(2,3))
print(arr)
```

```python
#生成随机数列(正态分布).两边小,中间大
arr=np.random.randn(2,3)
print(arr)
```

```python
#设置随机种子
np.random.seed(2000)
arr=np.random.randint(1,10,(2,5))
print(arr)
```

## 5.ndarray的索引与切片

```python
#一维数组索引与切片
arr=np.random.randint(1,100,15)
print(arr)
print(arr[0])
```

```python
print(arr[10])
print(arr[:])#全抽
print(arr[2:5])#抽取元素,左包右不包
print(arr[(arr>10) & (arr<70)])#布尔索引,&,|
print(arr[slice(2,5)])#与冒号作用相同

```

```python
#二维数组索引与切片
arr=np.random.randint(1,100,(3,4))
print(arr)
```

```python
print(arr[1,3])#2行4列
print(arr[1,1:3])#左行右列
print(arr[1][arr[1]>50])#布尔索引
print(arr[:,3])
```

## 5.ndarray的运算

```python
#算数的运算
a=np.array([[1,2,3],[2,4,6],[6,8,9]])
b=np.array([[4,5,6],[4,8,7],[1,6,9]])
print(a+3)#每个元素都加3
print(a-b)
print(a*b)
print(a/b)
```

```python
#广播机制(两个矩阵可以相乘为前提)
a=np.array([1,2,3])
b=np.array([[4],[5],[6]])
print(a+b)#一行与每一个第一行相加
print(b-a)
```

```python
a=np.array([[1,2,3],[2,4,6],[6,8,9]])
b=np.array([[4,5,6],[4,8,7],[1,6,9]])
print(b*a)
```

# 二.numpy常用函数
## 1.基本数学函数

```python
#计算平方根,np.sqrt
print(np.sqrt(9))
print(np.sqrt([1,4,9]))
arr=np.array([1,25,81])
print(np.sqrt(arr))
```

```python
#计算指数 np.exp,e的几次方
print(np.exp(1))
print(np.exp(0))
```

```python
#计算自然对数,以e为底
print(np.log(2.71))
```

```python
#计算正弦值,余弦值
print(np.sin(np.pi/2))
print(np.cos(np.pi))
```

```python
#计算绝对值,np.abs
arr=np.array([-1,1,2,-3])
print(np.abs(arr))
```

```python
#计算a的b次幂,np.power
print(np.power(arr,2))
```

```python
#四舍五入(五舍六入)
print(np.round([3.2,4.5,8.2,9.6]))
```

```python
#向上取整,向下取整,向上:ceil,向下:floor
arr=np.array([1.6,25.1,81.7])
print(np.ceil(arr))
print(np.floor(arr))
```

```python
#检测缺失值,np.isnan
np.isnan([1,2,np.nan,3])
```
