# 一.ndarray
## 1.ndarray的特性
### 多维性

```python
import numpy as np
```

```python
arr=np.array([1,2,3])
print(arr.ndim)#ndim指数组维度
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2.ndim)
print(arr2)
```

## 3.ndarray的创建

```python
#基础创建方法
list1=[4,5,6]
arr=np.array(list1,dtype=np.float64)#方法
print(arr.ndim)#属性
print(arr)
```

```python
#copy
arr1=np.copy(arr)#元素跟原始的数组相同,但不是一个数组
print(arr1)
arr1[0]=8
print(arr1)
print(arr)
```

```python
#预定义形状,全0,全1,未初始化,固定值
#全0
arr=np.zeros((2,3),dtype=int)
print(arr)
print(arr.dtype)
arr=np.zeros((2,),dtype=int)
print(arr)

#全1
arr=np.ones((2,1))
print(arr)
#未初始化
arr=np.empty((1,2))
print(arr)
arr=np.full((3,4),2025)
print(arr)
arr1=np.zeros_like(arr)
print(arr1)
```

```python
#等差数列
arr=np.arange(1,10,1)#start,end,step
print(arr)
```

```python
#等间隔数列
app=np.linspace(1,10,9)#start,end,num
print(app)
```

```python
#对数间隔数列
arr=np.logspace(0,4,3,base=2)#start,end,num,底数,底数的几次方
print(arr)
```

```python
#特殊矩阵
#单位矩阵:
arr=np.eye(3,4,dtype=int)#生成单位矩阵
print(arr)
```

```python
#对角矩阵
arr=np.diag([1,2,3])#生成对角矩阵
print(arr)
```

```python
#随机数组的生成
#生成0-1的随机浮点数(分布均匀)
arr=np.random.rand(2,3)
print(arr)
```

```python
#生成指定范围之间的随机浮点数
arr=np.random.uniform(3,6,(2,3))
print(arr)
```

```python
#生成随机整数
arr=np.random.randint(3,6,(2,3))
print(arr)
```

```python
#生成随机数列(正态分布).两边小,中间大
arr=np.random.randn(2,3)
print(arr)
```

```python
#设置随机种子
np.random.seed(2000)
arr=np.random.randint(1,10,(2,5))
print(arr)
```

## 5.ndarray的索引与切片

```python
#一维数组索引与切片
arr=np.random.randint(1,100,15)
print(arr)
print(arr[0])
```

```python
print(arr[10])
print(arr[:])#全抽
print(arr[2:5])#抽取元素,左包右不包
print(arr[(arr>10) & (arr<70)])#布尔索引,&,|
print(arr[slice(2,5)])#与冒号作用相同

```

```python
#二维数组索引与切片
arr=np.random.randint(1,100,(3,4))
print(arr)
```

```python
print(arr[1,3])#2行4列
print(arr[1,1:3])#左行右列
print(arr[1][arr[1]>50])#布尔索引
print(arr[:,3])
```

## 5.ndarray的运算

```python
#算数的运算
a=np.array([[1,2,3],[2,4,6],[6,8,9]])
b=np.array([[4,5,6],[4,8,7],[1,6,9]])
print(a+3)#每个元素都加3
print(a-b)
print(a*b)
print(a/b)
```

```python
#广播机制(两个矩阵可以相乘为前提)
a=np.array([1,2,3])
b=np.array([[4],[5],[6]])
print(a+b)#一行与每一个第一行相加
print(b-a)
```

```python
a=np.array([[1,2,3],[2,4,6],[6,8,9]])
b=np.array([[4,5,6],[4,8,7],[1,6,9]])
print(b*a)
```

```python
#求和,sum
print(np.sum(arr))
```

```python
#分位数
#中位数
print(np.median(arr))
np.random.seed(0)
arr=np.random.randint(0,100,4)
print(arr)
print(np.median(arr))
print(np.percentile(arr,20))#百分数分位
```

```python
#累计和,累积积
arr=np.array([1,2,3])
print(np.sum(arr))
print(np.cumsum(arr))#第三项为前两项的和,第四项为前三项的和
print(np.cumprod(arr))#累计同理

```

### 比较函数
比较函数
比较是否大于小于等于,逻辑与,或,非
检查数组中是否有一个true,是否所有的都为true

```python
#是否大于
print(np.greater([3,4,5,67,],4))#数组中的数与右侧数比较
#是否小于
print(np.less([3,4,5,67,8],4))
#是否等于
print(np.equal([3,4,5,6,7,8],4))
print(np.equal([1,2,3],[1,2,4]))#可比较,广播机制
```

```python
#与或非
print(np.logical_and([0,0],[1,1]))
print(np.logical_and([1,0],[1,1]))#判断前后对应的两个数是否为全1
print(np.logical_not([1,0]))#对一个元素进行判断
print(np.logical_or([0,0],[0,1]))
```

```python
#检查元素是否至少有一个元素为True
print(np.any([0,0,0]))
print(np.any([0,0,1]))
#所有元素为true
print(np.all([0,0,0]))
print(np.all([0,0,1]))
```

```python
#自定义条件
#print(np.where(条件,符合条件,不符合条件))
arr=np.array([1,2,3,4])
print(np.where(arr<3,arr,1))#不满足条件改为最后的数字
print(np.where(arr<3,1,0))
score=np.random.randint(1,100,10)
print(score)
print(np.where(score>=60,"及格","不及格"))
print(np.where(score<60,"不合格",np.where(score<80,"良好","优秀")))#嵌套,多类型
#np.select(条件,返回的结果)
print(np.select([score>80,(score>=60)&(score<=80),score<60],["优秀","良好","不及格"],default="未知"))
#default必须加,不然会报错
```

### 排序函数


```python
#排序函数
np.random.seed(0)
arr=np.random.randint(1,100,10)
print(arr)
print(np.sort(arr))#带上np,只输出排序后的数组,不会改变原数组
print(np.argsort(arr))#将索引排列输出
print(arr)
arr.sort()#会改变原数组
print(arr)
```

```python
#去重函数,np.unique,同时排序
print(np.unique(arr))
```

```python
#数组的拼接
arr1=np.array([1,2,34,])
arr2=np.array([4,5,7])
print(arr1+arr2)
print(np.concatenate((arr1,arr2)))
```

```python
#数组的分割
print(np.split(arr,10))#分多少组
print(np.split(arr,[3,5,7]))#分刀切
```

```python
#调整数组的形状,np.shape
print(np.reshape(arr,[2,5]))#改为几行几列
```
