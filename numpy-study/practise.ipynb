```python
#温度数据分析,某城市一周的最高气温为[28,30,29,31,32,30,29]
#计算平均气温,最高气温,最低气温
#找出气温超过30的天数
from http.client import TEMPORARY_REDIRECT

import numpy as np
tem=np.array([28,30,29,31,32,30,29])
print(f"平均气温:{np.mean(tem)}\n最高气温:{np.max(tem)}\n最低气温:{np.min(tem)}")
print("温度超过30天的天数",tem[tem>30].size)
print(np.sum(np.where(tem>30,1,0)))
print("温度超过30天的天数",len(tem[tem>30]))
print(np.count_nonzero(tem>30))
```

```python
"""学生成绩统计
五名同学成绩为[85,90,78,92,88]
计算成绩的平均分.中位数,标准差
将成绩转化为十分制"""
score=np.array([85,90,78,92,88])
print(f"中位数为:{np.median(score)}\n平均分为:{np.mean(score)}\n标准差为:{"%.3f"%np.std(score)}")
print(score/10)
```

```python
"""矩阵运算
给定矩阵A=[[1,2],[3,4]]和B=[[5,6],[7,8]]
计算A+B和A*B(逐元素乘法)
计算A和B的矩阵乘法(点积)"""
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
print(f"A+B={A+B}\nA*B={A*B}\nAB={A@B}")
```

```python
"""随机数据生成
生成一个(3,4)的随机整数数组,范围为[0,10)
计算每列最大值和每行最小值
将数组中的所有奇数替换为-1"""
np.random.seed(0)
arr1=np.random.randint(0,10,(3,4))
print(arr1)
for i in range(0,4):
    print(f"第{i+1}列最大值为:{np.max(arr1[:,i])}")#print(np.max(arr,axis=0)),0代表看列
for i in range(0,3):
    print(f"第{i+1}行最大值为{np.min(arr1[i,:])}")#print(np.max(arr,axis=1)),1代表看行

print(np.where([arr1%2!=0],-1,arr1))
#arr[arr1%2=1]=-1
#print(arr1)
```

```python
"""布尔索引
生成一个(5,5)的随机数组,范围为[0,20)
找出数组中大于10的元素
将所有大于10的元素替换为0"""
np.random.seed(0)
arr2=np.random.randint(0,20,(5,5))
print(arr2)
print(arr2[arr2>10])
# arr2[arr2>10]=0
#print(arr2)
print(np.where(arr2>10,0,arr2))
```

```python
"""数组变形
创建一个1-12的一维数组,并转换为(3,4)的二维数组
计算每行的和和每列的平均值
将数组展平为一维数组"""
np.random.seed(0)
arr3=np.random.randint(1,13,12)
print(arr3)
arr3=np.reshape(arr3,(3,4))
print(arr3)
print(f"每行的和:{np.sum(arr3,axis=1)}")
print(f"每列的平均值:{np.mean(arr3,axis=0)}")
print(np.reshape(arr3,(12)))
```

```python
"""统计函数的应用
公司六个月销售额为:[120,135,110,125,130,140]
计算销售额的总和,均值和方差
找出销售额最高的月份和最低的月份"""
money=np.array([120,135,110,125,130,140])
print(f"总和为:{np.sum(money)}\n均值为:{np.mean(money)}\n方差为:{np.var(money)}")
print(np.argmax(money)+1)
print(np.argmin(money)+1)
```

```python
"""数组拼接
给定A=[1,2,3]和B=[4,5,6]
将A与B水平拼接为一个新数组
将A与B垂直拼接为一个新数组"""
A=np.array([1,2,3])
B=np.array([4,5,6])
C=np.concatenate((A,B))
print(np.concatenate((A,B)))
print(np.reshape(C,(2,3)))
```

```python
"""缺失值处理
给定数组[1,np.nan,3,np.nan,5]
计算数组中的非缺失值的数量
将缺失值替换为零"""
a=np.array([1,np.nan,3,np.nan,5])
print(np.isnan(a))
print(np.sum(np.isnan(a)))
print(np.where(np.isnan(a),0,a))
```
