```python
"""
折线图 plot 趋势随时间变化
饼图 pie  整体组成比例
条形图 bar 类别之间对比
散点图 scatter 两变量相关性
箱线图 boxplot 数据分布,异常
多个图标
"""
```

```python
#绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'

#创建图表,设置大小
plt.figure(figsize=(10,5))#类似于画布
#显示图表
plt.show()
#绘制数据
month=["1月","2月","3月","4月"]
sales=[100,150,80,130]
plt.plot(month,sales,label='产品A')
#添加标题
plt.title('2025年销售趋势',color='red',fontsize=20)
#添加坐标轴的标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额(万元)',fontsize=10)
#添加图例
plt.legend(loc='upper left')
#添加网格线
# plt.grid(True)#x,y都有
plt.grid(axis='y',alpha=0.2,color='red',linestyle='--')#'-'指实线,'--'指虚线
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=15)#$前者是旋转角度,后者是坐标标签大小
#设置y周范围
plt.ylim(0,160)
#在每个数据点上显示数值
for x,y in zip(month,sales):

    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)
plt.show()
```
