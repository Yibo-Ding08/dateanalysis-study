```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'

#创建图表,设置大小
plt.figure(figsize=(10,5))#类似于画布
color=['red','blue','grey',"green",'yellow']
things=['学习','娱乐','运动',"睡觉","其他"]
times=[6,3,1,8,6]
plt.pie(times,labels=things,autopct='%1.f%%',startangle=90,colors=color,wedgeprops={'width':0.1})#调整初始绘画的角度)
plt.title('一天时间分布',color='red',fontsize=20)
plt.tight_layout()#自动优化
plt.show()
```

```python
#绘制爆炸式饼图

import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'

#创建图表,设置大小
plt.figure(figsize=(10,5))#类似于画布
color=['red','blue','grey',"green",'yellow']
explode=[0.5,0,0,0.7,0]#按位置来,谁对应非零,谁突出
things=['学习','娱乐','运动',"睡觉","其他"]
times=[6,3,1,8,6]
plt.pie(times,labels=things,autopct='%1.f%%',startangle=90,colors=color,explode=explode)#调整初始绘画的角度)
plt.title('一天时间分布',color='red',fontsize=20)
plt.tight_layout()#自动优化
plt.show()
```
