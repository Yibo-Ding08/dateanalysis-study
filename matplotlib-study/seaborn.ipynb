```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'
pg=pd.read_csv('E:\GitHub\date analys-study\pandas-study\Data\penguins.csv')
pg.head()

pg['sex']=pg['sex'].astype('category')
#直方图
sns.histplot(data=pg,x='species')


```

```python
"""核密度估计图"""
sns.histplot(data=pg,x='bill_length_mm',kde=True)#kde指的是在直方图上绘制线

```

```python
"""计数图"""
sns.countplot(data=pg,x='island')
```

```python
sns.histplot(data=pg,x='island')
```

```python
sns.scatterplot(data=pg,x='body_mass_g',
y='flipper_length_mm',hue='sex')#颜色按hue来
```

```python
#蜂窝图
sns.jointplot(data=pg,x='body_mass_g',y='flipper_length_mm',kind='hex')
```

```python
#箱线图
sns.boxplot(data=pg,x='species',y='bill_length_mm')
```

```python
#小提琴图
sns.pairplot(data=pg,hue='species')
```
