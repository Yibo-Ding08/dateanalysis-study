```python
#温度分析
import pandas as pd
import matplotlib .pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'
df=pd.read_csv('weather.csv')

#绘制气温的趋势
df['date']=pd.to_datetime(df['date'])
df=df[df['date'].dt.year==2012]
plt.figure(figsize=(5,5))
plt.title('2012年温度时间图',color='red')
plt.plot(df['date'],df['temp_max'],label='最高温')
plt.plot(df['date'],df['temp_min'],label='最低温')
plt.xlabel('日期',fontsize=20)
plt.ylabel('温度',fontsize=20)
# for x,y in zip (df['date'],df['temp_max']):
#     plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)
df['tem.mean']=(df['temp_max']+df['temp_min'])/2
plt.plot(df['date'],df['tem.mean'],label='平均气温')

plt.legend(loc='upper left')



```
