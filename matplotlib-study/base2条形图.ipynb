```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'

#创建图表,设置大小
plt.figure(figsize=(10,5))#类似于画布
sub=['语文','数学','英语','科学']
scores=[85,92,78,88]
#绘制柱状图
plt.bar(sub,scores,label='小红成绩',color='red',width=0.5)
plt.title('2025年成绩分布',color='red',fontsize=20)
plt.legend(loc='upper right')
plt.xlabel('科目',fontsize=10)
plt.ylabel('分数',fontsize=10)
plt.grid(axis='y',alpha=0.1,color='red',linestyle='--')
plt.ylim(0,100)
for x,y in zip(sub,scores):
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)

plt.tight_layout()#自动优化
plt.show()
```
