```python
#绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'

#创建图表,设置大小
plt.figure(figsize=(10,8))#类似于画布
#数据
scores=[50,55,65,70,79,80]
hours=[1,2,3,4,5,6]
plt.scatter(hours,scores)
plt.tight_layout()#自动优化
plt.show()
```

```python
#随机数
import random as rd
x=[]
y=[]
for i in range(100):
    tmp=rd.uniform(0,10)
    x.append(tmp)
    tmp2=2*tmp+rd.gauss(0,2)
    y.append(tmp2)
plt.scatter(x,y,alpha=0.5,s=20)#s为圆点大小
plt.title('x,y关系散点图',color='red',fontsize=20)
plt.legend(loc='upper right')
plt.xlabel('x',fontsize=20)
plt.ylabel('y',fontsize=20)
plt.grid(axis='y',alpha=0.1,color='red',linestyle='--')
plt.ylim(0,30)
plt.plot([0,10],[0,20],linewidth=1,color='red',linestyle="-")
plt.show()
```
